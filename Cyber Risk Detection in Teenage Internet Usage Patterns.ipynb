## **Problem Statement**



> The project aims to classify teenagers online behavior activity patterns into Safe, Neutral, or Risky cybersecurity behavior categories using machine learning by analyzing device usage, online behavior, security practices, and risk indicators. The model can help identify unsafe digital habits early and support strategies for promoting cybersecurity awareness among teenagers.



### **Type of Problem :** Classification

> **Input Features :** Timestamp, Device_Type,Malware_Detection, Phishing_Attempts,Social_Media_Usage, VPN_Usage, Cyberbullying_Reports,Parental_Control_Alerts, Firewall_Logs, Login_Attempts,Download_Risk, Password_Strength, Data_Breach_Notifications,Online_Purchase_Risk,Education_Content_Usage,Age_Group, Geolocation, Public_Network_Usage, Network_Type, Hours_Online, Website_Visits, Peer_Interactions, Risky_Website_Visits, Cloud_Service_Usage, Unencrypted_Traffic, Ad_Clicks, Insecure_Login_Attempts, E_Safety_Awareness_Score, Malware_Exposure_Risk.

> **Target Feature :** Cybersecurity_Behavior_Category

## **Objective**



> To develop a machine learning model that classifies teenage internet usage patterns into different cybersecurity behavior categories enabling early detection of risky online activities and promoting safer digital habits.






## **Outcomes**

> ♦  If a teenager uses a mobile device, spends 6 hours online, has weak passwords, clicks on ads frequently, and visits risky websites, the model predicts the outcome as **Risky**.

> ♦  If a teenager uses a laptop with low social media usage, strong parental control, and moderate safety awareness, the outcome is classified as **Safe**.

> ♦  If a teenager spending moderate hours online with weak passwords and occasional risky website visits is classified as **Neutral**.
